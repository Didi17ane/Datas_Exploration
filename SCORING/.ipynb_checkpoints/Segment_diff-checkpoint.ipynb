##### **Distribution Full Profiles**

```python
def pct_true(s):
    if s.dtype == "bool":
        return float(np.mean(s)) * 100.0
    try:
        return float(np.mean(s.astype(float))) * 100.0
    except Exception:
        return np.nan

# Variables numériques
numeric_vars = ['age_num']

# Variables catégorielles
categorical_vars = ['sex', 'marital_status', 'city', 'milieu_resid', 'region_name', 'bancarise']

profiles = []

for c in sorted(df_100['cluster'].unique()):
    sub = df_100[df_100['cluster'] == c]
    row = {
        'cluster': c,
        'taille': len(sub),
    }
    # Statistiques numériques
    for var in numeric_vars:
        row[f'{var}_median'] = sub[var].median()
        row[f'{var}_mean'] = sub[var].mean()
        row[f'{var}_std'] = sub[var].std()
        row[f'{var}_min'] = sub[var].min()
        row[f'{var}_max'] = sub[var].max()

    # Répartition pour catégorielles
    for var in categorical_vars:
        counts = sub[var].value_counts(dropna=False)
        total = counts.sum()
        repartition = {f'{var}_{cat}': count for cat, count in counts.items()}
        repartition_pct = {f'{var}_{cat}_pct': 100*count/total for cat, count in counts.items()}

        row.update(repartition)
        row.update(repartition_pct)

        # Modalité dominante
        mode_val = sub[var].mode(dropna=True)
        row[f'{var}_mode'] = mode_val.iloc[0] if not mode_val.empty else np.nan

    profiles.append(row)

profiles_df = pd.DataFrame(profiles).sort_values('cluster').reset_index(drop=True)

# Affichage synthétique
with pd.option_context('display.max_columns', None):
    print(profiles_df)

# Option : exporter vers CSV
profiles_df.to_csv("../DATAS/Distribution_full_profiles.csv", index=False)
```

#### Exemple 1 de Repartition des clusters

```python
# ------------------------------------------------------------
# 1. Génération des tranches d’âge
# ------------------------------------------------------------

def create_age_groups(df_100, col="age_num"):
    bins = [0,14,24,34,44,54,64,74,84,94,150]
    labels = [
        "0-14","15-24","25-34","35-44","45-54",
        "55-64","65-74","75-84","85-94","95+"
    ]
    df_100["age_group"] = pd.cut(df_100[col], bins=bins, labels=labels, right=True)
    return df_100


# ------------------------------------------------------------
# 2. Fonction d’analyse d’un cluster
# ------------------------------------------------------------

def analyze_single_cluster(sub, threshold=0.8):
    profile = {}

    # 2.1 — Statistiques numériques
    profile["age_mean"] = round(sub["age_num"].mean(),2)
    profile["age_median"] = round(sub["age_num"].median(),2)
    profile["age_std"] = round(sub["age_num"].std(),2)

    # 2.2 — Variables catégorielles à analyser
    categorical_vars = [
        "age_group","sex","marital_status","city",
        "milieu_resid","region_name","bancarise"
    ]

    dominant_rules = {}

    for var in categorical_vars:
        counts = sub[var].value_counts(normalize=True, dropna=False)

        # Modalités ≥ 80%
        dominant = counts[counts >= threshold]

        if len(dominant) > 0:
            dominant_rules[var] = dominant
        else:
            # S'il n'y a pas de modalité ≥ 80%, on garde la plus forte
            top = counts.iloc[0:1]
            dominant_rules[var] = top

    profile["dominant"] = dominant_rules
    return profile


# ------------------------------------------------------------
# 3. Génération d’un résumé textuel automatique
# ------------------------------------------------------------

def generate_summary(cluster_id, profile):
    dom = profile["dominant"]

    def format_dom(series):
        return ", ".join([f"{idx} ({round(val*100,2)}%)" for idx,val in series.items()])

    summary = f"Cluster {cluster_id} :\n"
    summary += f"- Âge moyen : {profile['age_mean']} ans (médiane {profile['age_median']})\n"
    summary += f"- Tranches d’âge dominantes : {format_dom(dom['age_group'])}\n"
    summary += f"- Sexe dominant : {format_dom(dom['sex'])}\n"
    summary += f"- Statut matrimonial dominant : {format_dom(dom['marital_status'])}\n"
    summary += f"- Milieu dominant : {format_dom(dom['milieu_resid'])}\n"
    summary += f"- Région dominante : {format_dom(dom['region_name'])}\n"
    summary += f"- Ville(s) dominante(s) : {format_dom(dom['city'])}\n"
    summary += f"- Bancarisation dominante : {format_dom(dom['bancarise'])}\n"

    return summary


# ------------------------------------------------------------
# 4. Fonction principale
# ------------------------------------------------------------

def generate_cluster_profiles(df_100, threshold=0.8):
    df_100 = df_100.copy()
    df_100 = create_age_groups(df_100)

    results = {}
    summaries = {}

    clusters = sorted(df_100["cluster"].unique())

    for c in clusters:
        sub = df_100[df_100["cluster"] == c]

        profile = analyze_single_cluster(sub, threshold)
        summary = generate_summary(c, profile)

        results[c] = profile
        summaries[c] = summary

        # Impression tableau propre
        print("\n" + "="*80)
        print(f"📊 PROFIL DU CLUSTER {c} (taille={len(sub)})")
        print("="*80)

        # Résumé texte
        print(summary)

        # Tableau des surreprésentations
        print("\n--- Modalités dominantes ---")
        for var, vals in profile["dominant"].items():
            print(f"\n{var} :")
            print(vals.apply(lambda x: round(x*100,2)))

    return results, summaries


# ------------------------------------------------------------
# Exemple d'utilisation
# ------------------------------------------------------------

results, summaries = generate_cluster_profiles(df_100, threshold=0.8)
```

```python
# -----------------------------------------------------------------------------
# 1) TABLEAU DES PROFILS NUMÉRIQUES PAR CLUSTER
# -----------------------------------------------------------------------------
def clusters_to_dataframe(results):
    rows = []

    for cluster_id, profile in results.items():
        row = {
            "cluster": cluster_id,
            "age_mean": profile["age_mean"],
            "age_median": profile["age_median"],
            "age_std": profile["age_std"],
        }
        rows.append(row)

    df_profiles = pd.DataFrame(rows)
    return df_profiles


# -----------------------------------------------------------------------------
# 2) TABLEAU DES MODALITÉS DOMINANTES POUR CHAQUE VARIABLE
# -----------------------------------------------------------------------------
def dominant_modalities_to_df(results):
    modal_rows = []

    for cluster_id, profile in results.items():
        dom = profile["dominant"]

        for var, series in dom.items():
            for modality, pct in series.items():
                modal_rows.append({
                    "cluster": cluster_id,
                    "variable": var,
                    "modalité": modality,
                    "pourcentage": round(pct * 100, 2)
                })

    df_modalities = pd.DataFrame(modal_rows)
    return df_modalities


# -----------------------------------------------------------------------------
# 3) TABLEAU DES RÉSUMÉS TEXTUELS
# -----------------------------------------------------------------------------
def summaries_to_df(summaries):
    rows = [{"cluster": cid, "résumé": text} for cid, text in summaries.items()]
    df_summaries = pd.DataFrame(rows)
    return df_summaries


# -----------------------------------------------------------------------------
# 4) EXÉCUTION : construire les 3 DataFrames finaux
# -----------------------------------------------------------------------------
df_profiles = clusters_to_dataframe(results)
df_modalities = dominant_modalities_to_df(results)
df_summaries = summaries_to_df(summaries)

# Aperçu
print(df_profiles.head())
print(df_modalities.head())
print(df_summaries.head())

# -----------------------------------------------------------------------------
# 5) EXPORT OPTIONNEL
# -----------------------------------------------------------------------------
#df_profiles.to_csv("cluster_profiles.csv", index=False)
#df_modalities.to_csv("cluster_modalities.csv", index=False)
#df_summaries.to_csv("cluster_summaries.csv", index=False)

print("=== Export terminé ===")
```

#### Exemple 2 de Repartition des clusters

```python
# Définition tranches d'âge
bins = [0, 14, 24, 34, 44, 54, 64, 74, 84, 94, 200]
labels = ["0-14", "15-24", "25-34", "35-44", "45-54", "55-64", "65-74", "75-84", "85-94", "95+"]
df_100['age_group'] = pd.cut(df_100['age_num'], bins=bins, labels=labels, right=True, include_lowest=True)

cat_vars = ['age_group', 'marital_status', 'milieu_resid', 'region_name', 'bancarise', 'sex', 'city']
seuil_presence = 0.8

def generate_cluster_profiles(df_100, cat_vars, seuil):
    for cluster_id in sorted(df_100['cluster'].unique()):
        sub = df_100[df_100['cluster'] == cluster_id]
        taille = len(sub)
        prop_cluster = taille / len(df_100)
        print(f"\n=== Cluster {cluster_id} - Taille : {taille} ({prop_cluster:.2%} pop totale) ===")
        
        profils = []
        
        for var in cat_vars:
            dist = sub[var].value_counts(normalize=True, dropna=False)
            # Modalités au-dessus du seuil
            above_seuil = dist[dist >= seuil]
            if not above_seuil.empty:
                for modalite, proportion in above_seuil.items():
                    profils.append(f"{proportion:.0%} dans '{modalite}' pour {var}")
            else:
                # Sinon afficher modalité la plus fréquente même si < seuil
                modalite_max = dist.idxmax()
                proportion_max = dist.max()
                profils.append(f"{proportion_max:.0%} dans '{modalite_max}' pour {var} (max)")
        
        if profils:
            print("Profil(s) dominant(s) ou maximal(aux) :")
            for p in profils:
                print(" - " + p)
        else:
            print("Pas de profil(s) détecté(s).")

# Exécution
generate_cluster_profiles(df_100, cat_vars, seuil_presence)
```

```python
# Préparation tranches d'âge
bins = [0,14,24,34,44,54,64,74,84,94,200]
labels = ["0-14","15-24","25-34","35-44","45-54","55-64","65-74","75-84","85-94","95+"]
df_100['age_group'] = pd.cut(df_100['age_num'], bins=bins, labels=labels, right=True, include_lowest=True)

cat_vars = ['age_group', 'marital_status', 'milieu_resid', 'region_name', 'bancarise', 'sex', 'city']
seuil_presence = 0.8

profiles_rows = []

for cluster_id in sorted(df_100['cluster'].unique()):
    sub = df_100[df_100['cluster'] == cluster_id]
    taille = len(sub)
    prop_cluster = taille / len(df_100)
    row = {'cluster': cluster_id, 'taille': taille, 'pct_population': prop_cluster}
    
    for var in cat_vars:
        dist = sub[var].value_counts(normalize=True, dropna=False)
        above_seuil = dist[dist >= seuil_presence]
        if not above_seuil.empty:
            # Liste modalités dominantes multiples si égales
            profils = [f"{modal} ({prop:.0%})" for modal, prop in above_seuil.items()]
        else:
            modalite_max = dist.idxmax()
            proportion_max = dist.max()
            profils = [f"{modalite_max} ({proportion_max:.0%}) (max)"]
        # On concatène ces descriptions en une seule chaîne par variable
        row[var+"_profil"] = "; ".join(profils)
    
    profiles_rows.append(row)

profiles_df = pd.DataFrame(profiles_rows).sort_values('cluster').reset_index(drop=True)

# Affichage ou export
print(profiles_df)

# Export CSV possible
profiles_df.to_csv("../DATAS/cluster_profiles_summary.csv", index=False)
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

# Tranches d'âge
bins = [0,14,24,34,44,54,64,74,84,94,200]
labels = ["0-14","15-24","25-34","35-44","45-54","55-64","65-74","75-84","85-94","95+"]
df_100['age_group'] = pd.cut(df_100['age_num'], bins=bins, labels=labels, right=True, include_lowest=True)

# Variables à encoder
cat_vars = ['age_group', 'marital_status', 'milieu_resid', 'region_name', 'bancarise', 'sex', 'city']
X = df_100[cat_vars]
y = df_100['cluster']

# Encodage one-hot
X_encoded = pd.get_dummies(X, drop_first=False)
```

```python
clf = DecisionTreeClassifier(max_depth=5, min_samples_leaf=20, random_state=42)
clf.fit(X_encoded, y)
```

```python
rules = export_text(clf, feature_names=list(X_encoded.columns))
print("Règles expliquant les clusters:\n")
print(rules)
```
