```python
cles = ["marital_status", "bancarise", "city", "milieu_resid", "age_num", "sex", "region_name"]

# Construire une clé combinée dans chaque df
clust["_key"] = clust[cles].astype(str).agg("||".join, axis=1)
concord["_key"] = concord[cles].astype(str).agg("||".join, axis=1)

# Marquer s'ils existent dans df2
concord["trouve"] = concord["_key"].isin(clust["_key"])
clust["trouve"] = clust["_key"].isin(concord["_key"])

# Résultat final : seulement les individus de df1
df_result1 = clust.drop(columns="_key")
df_result2 = concord.drop(columns="_key")
```

```python
keys = ["marital_status", "bancarise", "city", "milieu_resid", "age_num", "sex", "region_name"]

# s'assurer que les colonnes existent et sont comparables (str)
for c in keys:
    segm[c] = segm[c].astype(str)
    concordant[c] = concordant[c].astype(str)

# merge avec indicateur puis garder les "left_only" (présents dans segm mais pas dans concordant)
segm_not_in_concordant = segm.merge(
    concordant[keys].drop_duplicates(),
    on=keys,
    how="left",
    indicator=True
)
segm_not_in_concordant = segm_not_in_concordant[segm_not_in_concordant["_merge"] == "left_only"].drop(columns=["_merge"])
segm_not_in_concordant
```

```python
segm_not_in_concordant.columns
```

```python
segm_not_in_concordant = segm_not_in_concordant.drop(['agro_zone', 'department', 'hhweight',
       'resid', 'lien', 'religion', 'ethnie', 'nation', 'agemar', 'mal30j',
       'aff30j', 'arrmal', 'durarr', 'con30j', 'hos12m', 'couvmal', 'handit',
       'handig', 'alfa', 'alfa2', 'scol', 'educ_scol', 'educ_hi', 'diplome',
       'telpor', 'internet', 'activ7j', 'activ12m', 'branch', 'sectins', 'csp',
       'volhor', 'salaire', 'emploi_sec', 'sectins_sec', 'csp_sec',
       'volhor_sec', 'salaire_sec', 'serviceconsult', 'persconsult',
       'salaire_mois', 'salaire_sec_mois', 'rev_total_mois', 'log_revenu',
       'sans_revenu', 'age_grp', 'a_assurance', 'alphabete', 'logem',
       'elec_ac', 'elec_ur', 'elec_ua', 'tv', 'fer', 'frigo', 'cuisin',
       'ordin', 'decod', 'car', 'superf', 'volhor_manquant',
       'salaire_mois_manquant', 'Stg_formel'], axis=1)
segm_not_in_concordant
```

```python
# Grouper et compter les individus par cluster dans segm_not_in_concordant
cluster_counts = (
    segm_not_in_concordant['cluster']
    .value_counts(dropna=False)
    .rename_axis('cluster')
    .reset_index(name='count')
    .sort_values('count', ascending=False)
)

# ajouter pourcentage
cluster_counts['pct'] = (cluster_counts['count'] / cluster_counts['count'].sum()).round(4)

# afficher
cluster_counts
```
